# So what are the goals now?

So we would like a network to make decisions in the game that alters the state of the game, hence plays the game.

### Policy

This will be called the <b> policy </b> $\pi$, in full:

$$
\pi (a|s) = P(A_t=a|S_t=s)
$$

This policy makes decisions in the game based on the state. Wether decisions were good or bad can be quantified by a value called <b> return </b>, which is a cumulated sum of individual rewards:

### Return

$$
G_t = \sum_{k=0}^\infty\gamma^kR_{t+k}\;\;\;\textup{with   }\;\;\;R_t=r(a_t,s_t)
$$

Therefore, for $t=t'$, $R_{t'}$ will be weighted with 1 and $R_{t'+1}$ etc. will be weighted less. It should also be clear that for calculation of the return, the episode must have already been played out.

### Value function

Now given a policy $\pi$, one can ask: What is the expected value of a state $s$:

$$
V(s)=\mathbb{E}_\pi[G_t|S_t=s]
$$

This expectation goes quite a long way, considering that it must expect a sum, whichs "future parts" also depends on the policy that the expected value is calculated under...

### With the help of the idea of dynamic programming...

...it is possible to use the markov property to "simplify" this to the so called <b> Bellmann equation</b>:

$$
V(s)=\tilde{r}(s)+\gamma\sum_{s'\in S}P_\pi(s'|s)V(s')
$$

where

$$
P_\pi(s'|s) = \sum_{a\in A}\pi(a|s)P(s'|s,a)
$$

saying: "the transition probability is the product of where the policy wants to go and the chance that the sampled transition occurs"

so, written out:

$$
V(s)=\sum_{a\in A}\pi(a|s)\left[r(s,a)+\gamma\sum_{s'\in S}P(s'|s,a)V(s')\right]
$$

Now, the problem with the (potentially) infinite future states is lifted, but the way it is formulated, its hard to solve (can solve in matrix form, runs with complexity $n^3$ for n accessible states).

## The goal:

So the idea now is to find a policy that plays the game "optimally", creating a sequence of perfect actions given its observations. 

Formally, a policy is declared <b> optimal </b> if the following holds:

$$
V_\pi(s)\geq V_{\pi'} \;\;\forall s\in S\;\; \Rightarrow \pi\geq\pi'
$$

with the optimal value function

$$
V^*(s)=\textup{max}_{a\in A}\left[r(s,a)+\gamma\sum_{s'\in S} P(s'|s,a)V^*(s')\right]
$$

# Now, why not solve the Bellmann equations now?

The "tabular" Bellmann equations

$$
V^{\pi}(s)=\mathbb{E}_\pi[r+\gamma V^{\pi}(s')|s]=\mathbb{E}_\pi[G_t|S_t=s]
$$

We can now simply use the policy $\pi$ and gather many returns for the state s:

$$
V^{\pi}(s)=\mathbb{E}_\pi[G_t|S_t=s]\approx\frac{1}{N}\sum_{i=1}^NG_t^{(i)}
$$

IN THEORY, this would give a good approximation for the Value function if we do it for every state s.

However, the problem here is the <b> Sampling complexity</b>; In my game, there are far to many accessible states and it is infeasible to sample even a small portion of the total. 

What we need is a way to <b> approximate</b> the value of states by assuming that they share some kind of pattern in their value. This is the idea of <b> function approximation</b>; We parametrize the value function as a neural network, which learns to generalize to predict the approximated value of all states:

$$
V_\theta(s)
$$

Every deep RL algorithm still tries to satisfy the Bellmann equations; just approximately, over a learned representation.


However, now we have 2 <i>unknowns</i> we have to solve for; the value function $V_\theta(s)$ and the policy $\pi$ (which is also part of the value function lets not forget). How does one go on about that?

- <b> Generalized policy iteration </b>; rests on the [policy improvement theorem](https://www.tuananhle.co.uk/notes/policy-improvement-theorem.html): Essentially, switch back and forth between <b> policy iteration </b> (train policy network) and <b> policy evaluation </b> (train value approximation)
- <b> Actor-critic algorithm </b>; the critic network represents the action value function (Q function) approximation and the actor network represents the policy

# The difference and relation between the V- and Q function

They ask different questions:

- V asks: "How good is a state?". It does this by evaluating all possible (accessible) actions for state $s$:


$$
V(s)=\sum_{a\in A}\pi(a|s)\left[r(s,a)+\gamma\sum_{s'\in S}P(s'|s,a)V(s')\right]
$$

- The Q function asks an even more important questions (important as in when it comes to training the whole thing): "How good is the specific action a in a state s?". It relates to the V function in the following way:

$$
V(s)=\sum_{a\in A}\pi(a|s)Q(a,s)=\mathbb{E}_{a\sim\pi}[Q(a,s)]
$$


<i>Addon: Q-learning means to find $Q^*$ from which the best action can always greedily be sampled using argmax of action a. However for my game this is difficult, because its impossible to create a tabular fixed action space, but the action space is very dynamic.</i>

# So what is the training objective?

We want to approach this with a <b>policy-based</b> method, which stands opposite to <b>value-based</b> methods (Q-learning, deep Q Networks, ...) and involved modeling the policy $\pi$ as a neural network.

Specifically, in the actor-critic formulation, we have both a value network that is trained during policy evaluation and a policy network, which is trained during the policy iteration (ppo in my case)

The general training objective is to <b>maximize the expected return</b> from a starting state $s_0$:

$$
J(\theta)=\mathbb{E}_{s_0}\bigl[\mathbb{E}_{\pi_\theta}[G_t|S_t=s_0]\bigr]=\mathbb{E}_{s_0}[V^{\pi_\theta}(s_0)]
$$


.

.

.

.



## The gradient pulls out the Q function in the policy gradient formula:

$$
\nabla_\theta J(\theta)\propto \sum_{s\in S}\nu^{\pi_\theta}\sum_{a\in  A} Q^{\pi_\theta}(s,a)\nabla_\theta\pi_\theta(a|s) = \sum_{s\in S}\nu^{\pi_\theta}\sum_{a\in  A} Q^{\pi_\theta}(s,a) \pi_\theta(a|s)\frac{\nabla_\theta\pi_\theta(a|s)}{\pi_\theta(a|s)} = \mathbb{E}_{s\sim \nu^{\pi_\theta}, a\sim\pi(\bullet|s)}\biggl[Q^{\pi_\theta}(s,a)\nabla_\theta\bigl[\log\pi_\theta(a|s)\bigr]\biggr]
$$

The $\sum_{s\in S}\nu^{\pi_\theta}$ takes care of the fact, that only some distinct states have been visited from a specific initial state $s_0$. It is coincidentally called the <b>state-visitation function</b>. 

An <b>intuitive understanding</b>: The policy gradient formula consists of the gradient log term $\nabla_\theta\bigl[\log\pi_\theta(a|s)\bigr]\in\mathbb{R}^{|\theta|}$ which tells the algorithm how to update the parameters. And the Q-function ("how good is a specific action a in state s?") acts like a weight- higher values $Q(s,a)$ put more emphasis on the specific state-action tuple $(s,a)$.

## Why do we need an advantage function?

A common problem is that in a certain state, the Q-value does not necessarily depend on the action in the state so much as it depends on the state (if your in a winning position, many actions are valid, and if your in a loosing position already, then multiple action prob wont save your ass either...)

A far better idea is to exchange the (parametrized) action-value function with the <b>advantage function</b>:


$$
A(s,a)=Q(s,a)-V(s)
$$

This simultaneously means that $\sum_{a\in A}A(s_i,a)=1$ for a specific state $s_i$.

But still, we need to approximate the advantage function using the <b> temporal difference method</b>:

First, we need to make a crude assumption:

$$
\tilde Q(a_t,s_t)=r(a_t,s_t)+\gamma V(s_{t+1})
$$

This means that an action $a_t$ in state $s_t$ will deterministically lead to state $s_{t+1}$, effectively collapsing the sum to be just one term.

The <b> approximate advantage function </b> then becomes 

$$
\tilde A(a_t, s_t) = r_t +\gamma V(s_{t+1}) - V(s_t) = \delta_t
$$

Also remember that $V(s_t)$ is approximated by a neural network called the <i>critic</i>.

```python
## lets calculate the delta ts to see whats happening:
gamma = cfg.gamma

## ok first remember that we have to seperate the two players strands:
records    = list(collect_dict.values())
player_ids = np.array([sublist[-2] for sublist in records])

## rewards and values are now (2,) per step: column V_TERM and column V_DENSE,
## so every quantity below carries both streams side by side.
##
## Since sc-48 these deltas are finally meaningful. The critic used to be
## TRAINED on z-scored returns while its raw output was subtracted from raw
## rewards here and in the real GAE - a scale mismatch with no denormalisation
## anywhere, so `delta` mixed two different units. Both sides are raw return
## space now, and V_TERM is additionally bounded to [0, conquest_reward].
rewards_all = np.array([sublist[-5] for sublist in records], dtype=np.float64)
values_all  = np.stack([np.asarray(sublist[3].detach().cpu()) for sublist in records])

for p in [0,1]:
    p_steps: int = np.count_nonzero(player_ids==p)
    p_rewards  = rewards_all[player_ids==p]
    p_values   = values_all[player_ids==p]
    p_not_done = np.array([sublist[-4] for sublist in records])[player_ids==p]

    ## also add the last value per player to p_values:

    p_deltas = np.zeros((p_steps, 2))
    for i in range(p_steps-1, -1, -1):
        next_value = p_last_values[p] if i+1 == p_steps else p_values[i+1]
        p_deltas[i] = p_rewards[i] + gamma * next_value - p_values[i]

    
    if p==0:
        p0_bucket = (p_rewards, p_values, p_not_done, p_deltas)
    else:
        p1_bucket = (p_rewards, p_values, p_not_done, p_deltas)
```

## Now step it up a notch with the idea of <b>multi-step TD</b> and the <b>Generalized Advantage Estimate</b>

Now the deltas do hold some information above the single timestep t in form of the approximated estimate of the future state $V(s_{t+1})$, while also subtracting the value of the current state $V(s_t)$ as to really show the <b>advantage</b>.

Now, including multiple temporal difference steps gives rise to the following expressions:



Now the idea of the Generalized Advantage Estimation (GAE) is to balance the bias (single step TD) and variance (infinite step TD) of the multi-step advantage function by taking the weighted average:

$$
 A^{\textup{GAE}}_t=\frac{\sum_k w_k A^{(k)}_t}{\sum_k w_k}
$$

using exponential weights with $w_k=\lambda^{k-1}$ with $\lambda\in[0,1]$.

The GAE can be calculated in a simple iterative fashion:

$$
A^{\textup{GAE}}_t = \delta_t+\gamma\lambda A^{\textup{GAE}}_{t+1}
$$

starting from $t=T$ and moving all the way to $t=0$. The $t+1$ term is <b>bootstrapped</b> after the rollout horizon.

```python
p0_GAEs = np.zeros((len(p0_bucket[0]), 2))
p1_GAEs = np.zeros((len(p1_bucket[0]), 2))

_lambda = cfg.gae_lambda

advantage = np.zeros(2)
for i in range(len(p0_GAEs)-1, -1, -1):
    advantage = p0_bucket[-1][i] + gamma * _lambda * advantage
    p0_GAEs[i] = advantage

advantage = np.zeros(2)
for i in range(len(p1_GAEs)-1, -1, -1):
    advantage = p1_bucket[-1][i] + gamma * _lambda * advantage
    p1_GAEs[i] = advantage
```

```python
# GAE per player, per stream, plus the mix the policy actually optimises.
#
# The policy advantage is adv_term + dense_beta * adv_dense (batch_processing),
# so the third column is the only one that corresponds to a gradient. GAE is
# linear in the rewards, which is exactly why the two streams can be summed
# this way and why dense_beta is a clean dial rather than an approximation.
fig, axes = plt.subplots(2, 3, figsize=(18, 9))

beta = cfg.dense_beta
for row, (gaes, pid, c) in enumerate([(p0_GAEs, 0, "blue"), (p1_GAEs, 1, "red")]):
    mixed = gaes[:, V_TERM] + beta * gaes[:, V_DENSE]
    for col, (series, name) in enumerate([
        (gaes[:, V_TERM],  "terminal"),
        (gaes[:, V_DENSE], "dense"),
        (mixed,            f"mixed (beta={beta})"),
    ]):
        axes[row, col].scatter(np.arange(len(series)), series, color=c, s=18)
        axes[row, col].axhline(0.0, color="k", lw=0.6)
        axes[row, col].set_ylabel(f'GAE player {pid} ({name})')
        axes[row, col].set_xlabel("own decision index")
        axes[row, col].grid(alpha=0.25)
        axes[row, col].set_title(
            f"mean {series.mean():+.4f}   sd {series.std():.4f}", fontsize=9)

plt.tight_layout()
plt.show()

# PPO whitens the MIXED advantage across the batch before using it, so its sd is
# what sets the effective step size - worth watching alongside the raw values.
for pid, gaes in [(0, p0_GAEs), (1, p1_GAEs)]:
    mixed = gaes[:, V_TERM] + beta * gaes[:, V_DENSE]
    print(f"player {pid}: mixed advantage  mean {mixed.mean():+.5f}  "
          f"sd {mixed.std():.5f}  (term sd {gaes[:, V_TERM].std():.5f}, "
          f"dense sd {gaes[:, V_DENSE].std():.5f})")
```

## Inspecting a single decision's value distribution

`E[V_term]` is one number; the head predicts a whole distribution over `[0, conquest_reward]`. Two states with the same expectation can mean completely different things - a tight peak at 1.0 is "a close game, and I am sure of it", while mass split between 0 and 2 is "either I get conquered or I conquer". Only the second shape justifies a large advantage swing when the game resolves.


```python
from env.renderer import BoardRenderer

def show_value_dist(step: int):
    """Plot the critic's terminal-value distribution at one decision."""
    rep = dist_dict[step]
    fig, ax = plt.subplots(figsize=(7, 3.2))
    BoardRenderer(env).draw_value_distribution(
        ax, rep["term_probs"], rep["bin_values"],
        title=f"P(V_term) at decision {step}  (player {collect_dict[step][-2]})")
    plt.tight_layout(); plt.show()
    print(f"  E = {rep['term_value']:.4f}   mode = {rep['term_mode']:.3f}   "
          f"sd = {rep['term_std']:.4f}   V_dense = {rep['dense_value']:.4f}")

# The most and least confident decisions of the episode are usually the
# interesting ones to compare.
stds = np.array([dist_dict[i]["term_std"] for i in range(len(dist_dict))])
print(f"most confident decision:  {int(stds.argmin())}  (sd {stds.min():.4f})")
print(f"least confident decision: {int(stds.argmax())}  (sd {stds.max():.4f})")

show_value_dist(int(stds.argmin()))
show_value_dist(int(stds.argmax()))
```
